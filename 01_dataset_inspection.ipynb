````python
import numpy as np
import h5py
from scipy.io import loadmat
from pathlib import Path

BASE_DIR = Path(
    "/content/drive/MyDrive/Imagined_Speech_EEG_Project/dataset"
)

SPLIT_INFO = {
    "Training set": ("epo_train", "scipy"),
    "Validation set": ("epo_validation", "scipy"),
    "Test set": ("epo_test", "hdf5")
}

CLASS_NAMES = [
    "Hello",
    "Helpme",
    "Stop",
    "Thankyou",
    "Yes"
]


def load_eeg_file(file_path, split):

    variable_name, file_format = SPLIT_INFO[split]

    # Training and validation: MATLAB v5
    if file_format == "scipy":

        mat = loadmat(
            str(file_path),
            squeeze_me=True,
            struct_as_record=False
        )

        epo = mat[variable_name]

        # Original shape: time x channels x trials
        X = np.asarray(epo.x, dtype=np.float32)
        y_matrix = np.asarray(epo.y)

        fs = epo.fs
        channels = np.asarray(epo.clab)
        class_names = np.asarray(epo.className)

        # Convert to trials x channels x time
        X = np.transpose(X, (2, 1, 0))

        # Convert one-hot labels to class indices
        y = np.argmax(y_matrix, axis=0)

    # Test: MATLAB v7.3 / HDF5
    else:

        with h5py.File(file_path, "r") as f:

            group = f[variable_name]

            # HDF5 test shape: trials x channels x time
            X = np.asarray(
                group["x"],
                dtype=np.float32
            )

            # Read test labels as stored
            y_raw = np.asarray(group["y"])

            fs = float(group["fs"][0, 0])

            channels = np.asarray(group["clab"])
            class_names = np.asarray(group["className"])

            # Inspect labels before converting
            y = y_raw

    return X, y, fs, channels, class_names
````

````python
for split in SPLIT_INFO:

    file_path = (
        BASE_DIR /
        split /
        "Data_Sample01.mat"
    )

    X, y, fs, channels, class_names = load_eeg_file(
        file_path,
        split
    )

    print("\n==============================")
    print("SPLIT:", split)
    print("EEG shape:", X.shape)
    print("Label shape:", y.shape)
    print("Sampling rate:", fs)
    print("Number of channels:", len(channels))
    print("Class names:", class_names)
    print("Label values:", y)
````

````python

# ============================================================
# IMAGINED SPEECH EEG DATASET DOCUMENTATION
# Dataset inspection and descriptive statistics ONLY
# No preprocessing, no model training
# ============================================================

from pathlib import Path
from scipy.io import loadmat
import numpy as np
import pandas as pd
import h5py
import json
from datetime import datetime
from IPython.display import display, Markdown

# ------------------------------------------------------------
# 1. GOOGLE DRIVE PATHS
# ------------------------------------------------------------

from google.colab import drive
drive.mount('/content/drive')

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Imagined_Speech_EEG_Project"
)

DATASET_DIR = PROJECT_DIR / "dataset"

DOC_DIR = (
    PROJECT_DIR /
    "experiment_outputs" /
    "Dataset_Documentation"
)

DOC_DIR.mkdir(parents=True, exist_ok=True)

print("Dataset directory:", DATASET_DIR)
print("Documentation directory:", DOC_DIR)

# ------------------------------------------------------------
# 2. DATASET INFORMATION
# ------------------------------------------------------------

SPLITS = {
    "Training set": "epo_train",
    "Validation set": "epo_validation",
    "Test set": "epo_test"
}

CLASS_NAMES = [
    "Hello",
    "Helpme",
    "Stop",
    "Thankyou",
    "Yes"
]

NUM_CLASSES = len(CLASS_NAMES)

# ------------------------------------------------------------
# 3. MATLAB FILE LOADER FOR DOCUMENTATION
# ------------------------------------------------------------

def load_labelled_mat(file_path, variable):
    mat = loadmat(
        str(file_path),
        squeeze_me=True,
        struct_as_record=False
    )

    epo = mat[variable]

    X = np.asarray(epo.x)
    y_matrix = np.asarray(epo.y)

    labels = np.argmax(y_matrix, axis=0)

    channels = [
        str(x) for x in np.asarray(epo.clab).tolist()
    ]

    return {
        "shape_original": X.shape,
        "labels": labels,
        "sampling_rate": float(epo.fs),
        "channels": channels,
        "class_names": [
            str(x) for x in np.asarray(epo.className).tolist()
        ],
        "time": np.asarray(epo.t)
    }


def inspect_test_mat(file_path):
    with h5py.File(file_path, "r") as f:
        group = f["epo_test"]

        return {
            "shape_original": tuple(group["x"].shape),
            "sampling_rate": float(group["fs"][0, 0]),
            "label_shape": tuple(group["y"].shape),
            "label_dtype": str(group["y"].dtype),
            "label_raw": group["y"][()].tolist(),
            "channels_shape": tuple(group["clab"].shape),
            "class_names_shape": tuple(
                group["className"].shape
            )
        }


# ------------------------------------------------------------
# 4. INSPECT ALL SUBJECT FILES
# ------------------------------------------------------------

records = []
class_records = []
subject_records = []
test_records = []

for split_name, variable_name in SPLITS.items():

    folder = DATASET_DIR / split_name

    print("\nInspecting:", split_name)

    for file_path in sorted(folder.glob("*.mat")):

        subject = file_path.stem.replace("Data_Sample", "")

        if split_name in ["Training set", "Validation set"]:

            info = load_labelled_mat(
                file_path,
                variable_name
            )

            labels = info["labels"]
            counts = np.bincount(
                labels,
                minlength=NUM_CLASSES
            )

            # Standardized dimensions for labelled files
            original_shape = info["shape_original"]

            if len(original_shape) == 3:
                trials = len(labels)
                standardized_shape = (
                    trials,
                    len(info["channels"]),
                    len(info["time"])
                )
            else:
                standardized_shape = None

            records.append({
                "Split": split_name,
                "Subject": subject,
                "Trials": len(labels),
                "Channels": len(info["channels"]),
                "Time samples": len(info["time"]),
                "Sampling rate (Hz)": info["sampling_rate"],
                "Original EEG shape": str(original_shape),
                "Standardized EEG shape": str(
                    standardized_shape
                )
            })

            subject_records.append({
                "Split": split_name,
                "Subject": subject,
                "Total trials": len(labels),
                "Hello": int(counts[0]),
                "Helpme": int(counts[1]),
                "Stop": int(counts[2]),
                "Thankyou": int(counts[3]),
                "Yes": int(counts[4])
            })

            for i, class_name in enumerate(CLASS_NAMES):
                class_records.append({
                    "Split": split_name,
                    "Subject": subject,
                    "Class": class_name,
                    "Label": i,
                    "Trial count": int(counts[i])
                })

        else:
            # Test is HDF5 in the inspected dataset.
            info = inspect_test_mat(file_path)

            test_records.append({
                "Subject": subject,
                "EEG shape stored in HDF5": str(
                    info["shape_original"]
                ),
                "Sampling rate (Hz)": info["sampling_rate"],
                "y shape": str(info["label_shape"]),
                "y dtype": info["label_dtype"],
                "y raw values": str(info["label_raw"]),
                "Channel metadata shape": str(
                    info["channels_shape"]
                ),
                "Class metadata shape": str(
                    info["class_names_shape"]
                )
            })

# ------------------------------------------------------------
# 5. CREATE DATAFRAMES
# ------------------------------------------------------------

file_summary = pd.DataFrame(records)
subject_summary = pd.DataFrame(subject_records)
class_summary = pd.DataFrame(class_records)
test_summary = pd.DataFrame(test_records)

print("\nFILE SUMMARY")
display(file_summary)

print("\nSUBJECT-WISE CLASS DISTRIBUTION")
display(subject_summary)

print("\nTEST FILE STRUCTURE")
display(test_summary)

# ------------------------------------------------------------
# 6. OVERALL DATASET SUMMARY
# ------------------------------------------------------------

overall_rows = []

for split in ["Training set", "Validation set"]:
    part = file_summary[file_summary["Split"] == split]

    overall_rows.append({
        "Split": split,
        "Subjects inspected": part["Subject"].nunique(),
        "Total trials": int(part["Trials"].sum()),
        "Channels": sorted(part["Channels"].unique().tolist()),
        "Time samples": sorted(
            part["Time samples"].unique().tolist()
        ),
        "Sampling rate (Hz)": sorted(
            part["Sampling rate (Hz)"].unique().tolist()
        )
    })

overall_summary = pd.DataFrame(overall_rows)

print("\nOVERALL SUMMARY")
display(overall_summary)

# ------------------------------------------------------------
# 7. OVERALL CLASS DISTRIBUTION
# ------------------------------------------------------------

class_totals = (
    class_summary
    .groupby(["Split", "Class", "Label"])["Trial count"]
    .sum()
    .reset_index()
)

class_pivot = class_totals.pivot(
    index="Class",
    columns="Split",
    values="Trial count"
).reset_index()

print("\nOVERALL CLASS DISTRIBUTION")
display(class_pivot)

# ------------------------------------------------------------
# 8. ELECTRODE INFORMATION
# ------------------------------------------------------------

first_training_file = sorted(
    (DATASET_DIR / "Training set").glob("*.mat")
)[0]

first_info = load_labelled_mat(
    first_training_file,
    "epo_train"
)

electrode_table = pd.DataFrame({
    "Electrode Index": range(len(first_info["channels"])),
    "Electrode Name": first_info["channels"]
})

print("\nELECTRODE INFORMATION")
display(electrode_table)

# ------------------------------------------------------------
# 9. SAVE CSV TABLES
# ------------------------------------------------------------

file_summary.to_csv(
    DOC_DIR / "file_summary.csv",
    index=False
)

subject_summary.to_csv(
    DOC_DIR / "subject_class_distribution.csv",
    index=False
)

class_summary.to_csv(
    DOC_DIR / "class_distribution_by_subject.csv",
    index=False
)

class_pivot.to_csv(
    DOC_DIR / "overall_class_distribution.csv",
    index=False
)

electrode_table.to_csv(
    DOC_DIR / "electrode_information.csv",
    index=False
)

test_summary.to_csv(
    DOC_DIR / "test_file_structure.csv",
    index=False
)

# ------------------------------------------------------------
# 10. GENERATE DETAILED MARKDOWN DOCUMENTATION
# ------------------------------------------------------------

training_total = int(
    file_summary[
        file_summary["Split"] == "Training set"
    ]["Trials"].sum()
)

validation_total = int(
    file_summary[
        file_summary["Split"] == "Validation set"
    ]["Trials"].sum()
)

training_subjects = file_summary[
    file_summary["Split"] == "Training set"
]["Subject"].nunique()

validation_subjects = file_summary[
    file_summary["Split"] == "Validation set"
]["Subject"].nunique()

report = f"""
# Imagined Speech EEG Dataset
## Dataset Description and Contents

**Documentation date:** {datetime.now().strftime("%d %B %Y")}

---

## 1. Dataset Overview

The dataset used in this project is the Imagined Speech EEG
dataset associated with the 2020 International BCI Competition,
Track 3.

**Dataset source:** https://osf.io/pq7vb/

The dataset contains EEG recordings corresponding to five
imagined speech classes.

The dataset information supplied for this project describes
15 participants, identified as S1–S15, aged 20–30 years.

## 2. Imagined Speech Classes

| Label | Class name |
|---:|---|
| 0 | Hello |
| 1 | Helpme |
| 2 | Stop |
| 3 | Thankyou |
| 4 | Yes |

Total number of classes: **5**

## 3. Dataset Folder Structure

```text
Imagined_Speech_EEG_Project/
|
|-- dataset/
|   |
|   |-- Training set/
|   |   |-- Data_Sample01.mat
|   |   |-- ...
|   |   |-- Data_Sample15.mat
|   |
|   |-- Validation set/
|   |   |-- Data_Sample01.mat
|   |   |-- ...
|   |   |-- Data_Sample15.mat
|   |
|   |-- Test set/
|       |-- Data_Sample01.mat
|       |-- ...
|       |-- Data_Sample15.mat
|
|-- experiment_outputs/
    |-- Dataset_Documentation/
```

## 4. EEG Signal Characteristics

| Parameter | Value |
|---|---:|
| Sampling frequency | 256 Hz |
| Number of EEG channels | 64 |
| Samples per trial | 795 |
| Approximate duration | {795/256:.4f} seconds |
| Number of classes | 5 |
| Canonical labelled data layout | Trials × Channels × Time |

The Training and Validation files were converted from their
stored MATLAB array order to the canonical order:

`(trials, channels, time_samples)`.

## 5. Dataset Split Summary

| Split | Subjects inspected | Total trials |
|---|---:|---:|
| Training set | {training_subjects} | {training_total} |
| Validation set | {validation_subjects} | {validation_total} |
| Test set | {len(test_summary)} | Not inferred from labels |

## 6. Training Set Class Distribution

| Class | Trials per subject | Total trials |
|---|---:|---:|
| Hello | 60 | {60*training_subjects} |
| Helpme | 60 | {60*training_subjects} |
| Stop | 60 | {60*training_subjects} |
| Thankyou | 60 | {60*training_subjects} |
| Yes | 60 | {60*training_subjects} |

## 7. Validation Set Class Distribution

| Class | Trials per subject | Total trials |
|---|---:|---:|
| Hello | 10 | {10*validation_subjects} |
| Helpme | 10 | {10*validation_subjects} |
| Stop | 10 | {10*validation_subjects} |
| Thankyou | 10 | {10*validation_subjects} |
| Yes | 10 | {10*validation_subjects} |

## 8. Electrode Information

The inspected Training metadata provides the following
{len(first_info['channels'])} electrode names:

{', '.join(first_info['channels'])}

The complete electrode index and name mapping is available
in `electrode_information.csv`.

## 9. Subject-wise Dataset Contents

The file `subject_class_distribution.csv` records the number
of trials for each class in every inspected subject and split.

The file `file_summary.csv` records the dimensions and
sampling frequency for every inspected labelled file.

## 10. Test Dataset File Structure

The inspected Test files use MATLAB v7.3/HDF5 format.

For the first inspected Test file:

- EEG dataset shape: (50, 64, 795)
- Sampling frequency: 256 Hz
- `epo_test/y` shape: (2,)
- `epo_test/y` data type: uint64
- Raw `epo_test/y` values: [0, 0]

The two-element `y` field is not a per-trial label vector
for the 50 EEG trials. Its meaning has not been established.
The test-label contents must therefore not be interpreted
as class labels.

See `test_file_structure.csv` for the structural inspection
of the Test files.

## 11. Dataset Balance

The Training and Validation class distributions were
calculated from the labels in all inspected subject files.

The Training set contains 60 trials per class per subject.
The Validation set contains 10 trials per class per subject.

## 12. MATLAB File Contents

The inspected Training MATLAB variable is `epo_train`.
Its fields include:

- `x`: EEG signal array
- `y`: class label matrix
- `fs`: sampling frequency
- `clab`: channel labels
- `className`: class names
- `t`: time vector
- `title`: subject title
- `file`: file information

The inspected Validation variable is `epo_validation`
with corresponding EEG and metadata fields.

The inspected Test HDF5 group is `epo_test`.

## 13. Documentation Scope

This document describes the dataset and its contents only.

It does not contain:
- EEG preprocessing results
- Wavelet or scalogram results
- Model training
- Classification accuracy
- Model comparisons
- Experimental conclusions

---

**End of Dataset Documentation**
"""

md_file = DOC_DIR / "Dataset_Documentation_Report.md"

with open(md_file, "w", encoding="utf-8") as f:
    f.write(report)

# ------------------------------------------------------------
# 11. SAVE MACHINE-READABLE DATASET METADATA
# ------------------------------------------------------------

metadata = {
    "dataset_name": "Imagined Speech EEG - BCI Competition 2020 Track 3",
    "source": "https://osf.io/pq7vb/",
    "classes": CLASS_NAMES,
    "number_of_classes": NUM_CLASSES,
    "sampling_rate_hz": 256,
    "channels": first_info["channels"],
    "time_samples_per_trial": 795,
    "training_trials_verified": training_total,
    "validation_trials_verified": validation_total,
    "training_subjects_verified": training_subjects,
    "validation_subjects_verified": validation_subjects,
    "test_files_inspected": len(test_summary),
    "documentation_scope": "Dataset description only"
}

with open(DOC_DIR / "dataset_metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

# ------------------------------------------------------------
# 12. DISPLAY REPORT AND FILE LIST
# ------------------------------------------------------------

display(Markdown(report))

print("\nDOCUMENTATION GENERATED SUCCESSFULLY")
print("=" * 55)

for file in sorted(DOC_DIR.iterdir()):
    print(file.name, "-", round(file.stat().st_size / 1024, 2), "KB")
````
